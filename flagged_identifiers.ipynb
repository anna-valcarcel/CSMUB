```python
import shutil
import os
import pandas as pd
```

```python
masterfile = "/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CSMUB/RESULTS/DELINEATION_TEST/test_delineation_stats_FINAL.csv"
df = pd.read_csv(masterfile)

print(df.columns)
```

```python
indx = df.index[df['FLAG_small_Intersec<80'] == True].tolist()
print(len(indx))
```

```python
stations = df['Station'][indx]
station_nums = [str(x) for x in stations]
```

```python
source_folder_path = "/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CSMUB/RESULTS/DELINEATION_TEST/"  # Replace with your source folder path
destination_folder_path = "/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CSMUB/RESULTS/DELINEATION_TEST/small_intersection/"  # Replace with your destination folder path

# Ensure the destination folder exists
os.makedirs(destination_folder_path, exist_ok=True)

# Iterate through files in the source folder and copy them
for station in station_nums:
    for filename in os.listdir(source_folder_path):
        if station in filename:
            source_file_path = os.path.join(source_folder_path, filename)
            destination_file_path = os.path.join(destination_folder_path, filename)
    
            # Check if it's a file before copying (to avoid copying subdirectories)
            if os.path.isfile(source_file_path):
                shutil.copy(source_file_path, destination_file_path)
                print(f"File '{filename}' copied to '{destination_folder_path}'")
```
