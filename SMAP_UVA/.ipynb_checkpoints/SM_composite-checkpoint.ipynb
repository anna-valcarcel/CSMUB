### Code to average SM over 4 weekly files

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import pandas as pd
import rioxarray
import time
from pyproj import Transformer
import os
#import sys
```

```python
# Use transformer to transform coords
transformer = Transformer.from_crs("epsg:4326","epsg:6933",  always_xy=True)

# corner coords - CYGNSS
max_lat_EASE = 38.834339439960516
max_lon_EASE = -127.07799119974123
min_lat_EASE = -38.05351743979274
min_lon_EASE = 154.52355987593953

# corner coords - temporary Tunisia
'''
max_lat_EASE = 32.756257076586365
max_lon_EASE = 9.82531256160529
min_lat_EASE = 24.21130659519007
min_lon_EASE = 22.7623218320786
'''

# Transform
max_x, max_y = transformer.transform(max_lon_EASE, max_lat_EASE)
min_x, min_y = transformer.transform(min_lon_EASE, min_lat_EASE)
print(f"Longitude: {max_x}, Latitude: {max_y}")
print(f"Longitude: {min_x}, Latitude: {min_y}")

# delete cache
vars_to_delete = ["max_lat_EASE", "min_lat_EASE", "max_lon_EASE", "min_lon_EASE", "Transformer", "transformer"]
for v in vars_to_delete:
    if v in globals():
        del globals()[v]
```

```python
# Code 1 - Float 16 (WORKS)
def composite_rasters(fns, max_x, max_y, min_x, min_y, dtype=np.float16):
    start_time = time.perf_counter()
    
    avg = None
    count = None
    template = None
    
    for i, fn in enumerate(fns, start=1):
        t0 = time.perf_counter()
        print(f'  Processing file {i}: {os.path.basename(fn)}')
        
        # Open raster with chunking
        ds = rioxarray.open_rasterio(fn, chunks={"x": 5000, "y": 5000})
        ds_crop = ds.sel(x=slice(max_x, min_x), y=slice(max_y, min_y))

        # Save template
        if template is None:
            template = ds_crop.isel(band=0)
        
        # Convert to desired dtype
        arr = ds_crop.isel(band=0).astype(dtype).values
        
        # Initialize avg and count on first array
        if avg is None:
            avg = np.zeros_like(arr, dtype=np.float16)
            count = np.zeros_like(arr, dtype=np.int32)
        
        # Incremental mean: add valid pixels
        mask = ~np.isnan(arr)
        avg[mask] += arr[mask]
        count[mask] += 1
        
        elapsed = time.perf_counter() - t0
        print(f'  File {i} done, time: {elapsed:.2f}s')
        
        # Explicitly delete arr and ds_crop to free memory
        del arr, ds_crop, ds
    
    # Compute final mean
    avg[count > 0] /= count[count > 0]
    avg[count == 0] = np.nan
    
    print(f'  Wrapping back to .nc...')

    # Wrap into xr dataframe to return
    avg_da = xr.DataArray(
        avg,
        dims=template.dims,
        coords=template.coords,
        name="composite"
    )

    avg_da.rio.write_crs(template.rio.crs, inplace=True)
    avg_da.rio.write_transform(template.rio.transform(), inplace=True)

    total_time = time.perf_counter() - start_time
    print(f' >TOTAL COMOSITE TIME = {total_time:.2f}s')
    
    return avg_da

def save_composite(month, year):
    # get the fns of the weeks
    dir = f'/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CYGNSS/Data/sls_SOILMOISTURE_UVA/smap_sm_400m_{year}_week_'
    month = 1
    weeks = np.arange(month*4-3, month*4+1)
    weeks = [f"{w:02d}" for w in weeks]
    print(f'\nMONTH : {month} ------ WEEKS: {weeks} ------ YEAR: {year}')
    fns = []
    for week in weeks:
        fn = f'{dir}{week}.tif'
        fns.append(fn)
    
    # Composite
    avg_da = composite_rasters(fns, max_x, max_y, min_x, min_y, dtype=np.float16)

    # Save to ds
    save_dir = '/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CYGNSS/Data/sls_SOILMOISTURE_UVA/monthly_composite/'
    #avg_da.to_netcdf(f"{save_dir}test.nc")
    return(avg_da)

month = 1
year = 2015
avg_test_1 = save_composite(month, year)
```

```python
# Code 3 - with GC (Wrap extra week) (WORKS) - ran 1x
import numpy as np
import xarray as xr
import rioxarray
import os, time, gc

def get_weeks_for_month(month):
    # weeks starting
    month_start = [1, 5, 9, 13, 17, 21, 25, 29, 33, 37, 41, 45]
    
    # Base 4-week range
    start = month_start[month - 1]
    weeks = list(range(start, start + 4))
    
    # distribute
    extra_weeks = [49, 50, 51, 52]
    long_months = [3, 5, 8, 10]
    if month in long_months:
        weeks.append(extra_weeks.pop(0))
    
    return weeks

# -----------------------------------
# Composite function
# -----------------------------------
def composite_rasters(fns, max_x, max_y, min_x, min_y, dtype=np.float32):
    start_time = time.perf_counter()
    
    avg = None
    count = None
    template = None
    
    for i, fn in enumerate(fns, start=1):
        t0 = time.perf_counter()
        print(f'  Processing file {i}: {os.path.basename(fn)}')
        
        ds = rioxarray.open_rasterio(fn, chunks={"x": 5000, "y": 5000})
        ds_crop = ds.sel(x=slice(max_x, min_x), y=slice(max_y, min_y))

        if template is None:
            template = ds_crop.isel(band=0)
        
        arr = ds_crop.isel(band=0).astype(dtype).values
        
        if avg is None:
            avg = np.zeros_like(arr, dtype=np.float32)
            count = np.zeros_like(arr, dtype=np.int32)
        
        mask = ~np.isnan(arr)
        avg[mask] += arr[mask]
        count[mask] += 1
        
        print(f'  File {i} done, time: {time.perf_counter() - t0:.2f}s')
        
        # NEW: gc
        del arr, ds_crop
        ds.close()
        del ds
        gc.collect()   # helps prevent buildup
    
    # Final mean
    avg[count > 0] /= count[count > 0]
    avg[count == 0] = np.nan
    
    print('  Wrapping back to .nc...')

    avg_da = xr.DataArray(
        avg,
        dims=template.dims,
        coords=template.coords,
        name="composite"
    )

    avg_da.rio.write_crs(template.rio.crs, inplace=True)
    avg_da.rio.write_transform(template.rio.transform(), inplace=True)

    print(f' >TOTAL COMPOSITE TIME = {time.perf_counter() - start_time:.2f}s')
    
    return avg_da


# -----------------------------------
# Save function
# -----------------------------------
def save_composite(month, year):
    base_dir = f'/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CYGNSS/Data/sls_SOILMOISTURE_UVA/smap_sm_400m_{year}_week_'
    
    #weeks = np.arange(month*4-3, month*4+1)
    #weeks = [f"{w:02d}" for w in weeks]
    
    # distribution!!!
    weeks = [f"{w:02d}" for w in get_weeks_for_month(month)]
    
    print(f'\nMONTH: {month} | WEEKS: {weeks} | YEAR: {year}')
    
    fns = [f"{base_dir}{week}.tif" for week in weeks]
    
    avg_da = composite_rasters(fns, max_x, max_y, min_x, min_y)

    save_dir = '/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CYGNSS/Data/sls_SOILMOISTURE_UVA/monthly_composite/'
    out_fn = f"{save_dir}{year}_{month}_composite_extraMonths.nc"
    
    avg_da.to_netcdf(out_fn)
    print(f" >SAVED TO: {out_fn}")
    
    # NEW Clean up after each month
    avg_da.close()
    del avg_da
    gc.collect()


# -----------------------------------
# Run loop
# -----------------------------------
months = np.arange(7, 13)
years = np.arange(2017, 2025)
for year in years:
    for month in months:
        try:
            save_composite(month, year)
            gc.collect()
        except: 
            print('FAILED MONTH!!! D:')
```

```python
# Code 4 - weight the weeks ( DEBUGGER)
import numpy as np
import xarray as xr
import rioxarray
import os, time, gc
from datetime import datetime, timedelta
from collections import defaultdict

def week_to_dates(year, week):
    start = datetime.fromisocalendar(year, week, 1)  # Monday
    return [start + timedelta(days=i) for i in range(7)]

def compute_month_week_weights(year):

    month_map = defaultdict(list)

    for week in range(1, 53):
        try:
            dates = week_to_dates(year, week)
        except ValueError:
            continue 

        # counting days per month
        month_counts = defaultdict(int)
        for d in dates:
            if d.year == year:  
                month_counts[d.month] += 1

        # wights
        for month, days in month_counts.items():
            weight = days / 7.0
            month_map[month].append((week, weight))

    return month_map

# -----------------------------------
# Composite function (NOW WEIGHTED)
# -----------------------------------
def composite_rasters_weighted(week_weights, year, max_x, max_y, min_x, min_y, dtype=np.float32):
    start_time = time.perf_counter()

    avg = None
    count = None
    template = None

    base_dir = f'/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CYGNSS/Data/sls_SOILMOISTURE_UVA/smap_sm_400m_{year}_week_'

    for i, (week, weight) in enumerate(week_weights, start=1):
        fn = f"{base_dir}{week:02d}.tif"

        if not os.path.exists(fn):
            print(f"  Missing file: {fn}")
            continue

        t0 = time.perf_counter()
        print(f'  Processing week {week} (weight={weight:.2f})')

        ds = rioxarray.open_rasterio(fn, chunks={"x": 5000, "y": 5000})
        ds_crop = ds.sel(x=slice(max_x, min_x), y=slice(max_y, min_y))
        
        if template is None:
            template = ds_crop.isel(band=0)

        arr = ds_crop.isel(band=0).astype(dtype).values

        #PLOT - debugging ---------------------
        plt.figure()
        plt.imshow(arr)
        plt.title(f"Year {year} Week {week} (weight={weight:.2f})")
        plt.colorbar()
        plt.show()
        plt.close()
        print(f"Value at (1000,1000): {arr[1000, 1000]}")
        #--------------------

        if avg is None:
            avg = np.zeros_like(arr, dtype=np.float32)
            count = np.zeros_like(arr, dtype=np.float32)  # NOTE: float now!

        mask = ~np.isnan(arr)

        # Weighted accum... (NEW)
        avg[mask] += weight * arr[mask]
        count[mask] += weight

        print(f'  Done week {week}, time: {time.perf_counter() - t0:.2f}s')

        # cleanup
        del arr, ds_crop
        ds.close()
        del ds
        gc.collect()

    # Final weighted mean
    avg[count > 0] /= count[count > 0]
    avg[count == 0] = np.nan

    print('  Wrapping back to .nc...')

    avg_da = xr.DataArray(
        avg,
        dims=template.dims,
        coords=template.coords,
        name="composite"
    )

    avg_da.rio.write_crs(template.rio.crs, inplace=True)
    avg_da.rio.write_transform(template.rio.transform(), inplace=True)

    print(f' >TOTAL COMPOSITE TIME = {time.perf_counter() - start_time:.2f}s')

    return avg_da


# -----------------------------------
# Save function
# -----------------------------------
def save_composite(month, year, month_map):
    print(f'\nMONTH: {month} | YEAR: {year}')

    week_weights = month_map[month]

    avg_da = composite_rasters_weighted(
        week_weights,
        year,
        max_x, max_y, min_x, min_y
    )

    save_dir = '/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CYGNSS/Data/sls_SOILMOISTURE_UVA/monthly_composite/'
    out_fn = f"{save_dir}{year}_{month}_composite_weighted.nc"

    #avg_da.to_netcdf(out_fn)
    print(f" >SAVED TO: {out_fn}")
    return(avg_da)
    #avg_da.close()
    #del avg_da
    #gc.collect()


# -----------------------------------
# Run loop
# -----------------------------------
months = np.arange(1, 13)
years = np.arange(2021, 2023)

for year in years:
    print(f"\n==== YEAR {year} ====")

    # Precompute weights once per year
    month_map = compute_month_week_weights(year)

    for month in months:
        try:
            test = save_composite(month, year, month_map)
        except Exception as e:
            print(f'FAILED MONTH {month} {year}: {e}')
            gc.collect()
```

```python
# Use transformer to transform coords
from pyproj import Transformer
# ease grid tansformer
transformer = Transformer.from_crs("epsg:4326","epsg:6933",  always_xy=True)

# corner coords - CYGNSS
#max_lat_EASE = 38.834339439960516
#max_lon_EASE = -127.07799119974123
#min_lat_EASE = -38.05351743979274
#min_lon_EASE = 154.52355987593953

# corner coords - temporary Tunisia
max_lat_EASE = 32.756257076586365
max_lon_EASE = 9.82531256160529
min_lat_EASE = 24.21130659519007
min_lon_EASE = 22.7623218320786

# Transform
max_x, max_y = transformer.transform(max_lon_EASE, max_lat_EASE)
min_x, min_y = transformer.transform(min_lon_EASE, min_lat_EASE)
print(f"Longitude: {max_x}, Latitude: {max_y}")
print(f"Longitude: {min_x}, Latitude: {min_y}")

# delete cache
vars_to_delete = ["max_lat_EASE", "min_lat_EASE", "max_lon_EASE", "min_lon_EASE", "Transformer", "transformer"]
for v in vars_to_delete:
    if v in globals():
        del globals()[v]
```

```python
def reduce_ds(fn, max_x, max_y, min_x, min_y):
    # Open Dataset
    ds = rioxarray.open_rasterio(fn, chunks={"x": 5000, "y": 5000})
    
    # Crop
    ds_crop = ds.sel(
        x=slice(max_x, min_x),
        y=slice(max_y, min_y)  
    )
    
    # Convert to float 16
    arr_16 = ds_crop.isel(band=0)#.astype(np.float16)
    
    # Summary of reduction
    #idx_x, idx_y = 0,0
    print(f'Original size: {ds.nbytes/(10**9)} GB')
    print(f'Cropped size: {ds_crop.nbytes/(10**9)} GB')
    print(f'convert to float16 size: {arr_16.nbytes/(10**9)} GB')
    #print(f'numpy array: {arr_16_np[idx_x,idx_y]} / size: {arr_16_np.nbytes} bytes')
    print(f'---- TOTAL REDUCTION = {arr_16.nbytes/ds.nbytes}% of original -----\n')
    
    # Clear CacheS
    return(arr_16)
 
```

```python
arrs = []
arrs_ds = []
for fn in fns:
    arr_16 = reduce_ds(fn, max_x, max_y, min_x, min_y)
    arrs_ds.append(arr_16)
    arrs.append(arr_16.values)
arrs
```

#### Compositing

```python
# With XR
def XR_composite(arr_list):
    data_concat = xr.concat(arr_list, dim = "stack")
    print(f"concatenated, {data_concat.nbytes/(10**9)} GB")
    avg = data_concat.mean("stack", skipna=True, dtype = "float16").compute()
    print(f"Averaged, {avg.nbytes/(10**9)} GB")
    return(avg)

#start_time = time.perf_counter()
#avg = XR_composite(arrs)
#end_time = time.perf_counter()
#elapsed_time = end_time - start_time
#print(f"Elapsed time: {elapsed_time:.4f} seconds")
```

```python
# Incremental 1 - BAD
def INC_composite(arr_list):
    # Initialize the arrs with arr 1
    start_time = time.perf_counter()
    avg = arr_list[0]
    count = (~np.isnan(avg)).astype(np.int32)
    elapsed_time = time.perf_counter() - start_time
    print(f'Copied first array & initialized count, time: {elapsed_time:.2f}s')
    
    # Loop remaining arrs
    for i, arr in enumerate(arr_list[1:], start=2): # starting with 2...
        start_time_2 = time.perf_counter()
        print(f'Running array {i}')
        mask = ~np.isnan(arr) # nan mask for count
        avg[mask] += arr[mask]
        count[mask] += 1
        elapsed_time2 = time.perf_counter() - start_time_2
        print(f'Array {i} complete, time: {elapsed_time2:.2f}s')
    
    # Composite / compute mean
    avg[count > 0] /= count[count > 0]
    avg[count == 0] = np.nan
    
    elapsed_time = time.perf_counter() - start_time
    print(f'TOTAL TIME = {elapsed_time:.2f}s')
    
    return avg
```

```python
# Incremental 2 
import os
def composite_rasters(fns, max_x, max_y, min_x, min_y, dtype=np.float16):
    start_time = time.perf_counter()
    
    avg = None
    count = None
    
    for i, fn in enumerate(fns, start=1):
        t0 = time.perf_counter()
        print(f'Processing file {i}: {os.path.basename(fn)}')
        
        # Open raster with chunking
        ds = rioxarray.open_rasterio(fn, chunks={"x": 5000, "y": 5000})
        ds_crop = ds.sel(x=slice(max_x, min_x), y=slice(max_y, min_y))
        
        # Convert to desired dtype
        arr = ds_crop.isel(band=0).astype(dtype).values
        
        # Initialize avg and count on first array
        if avg is None:
            avg = np.zeros_like(arr, dtype=np.float16)
            count = np.zeros_like(arr, dtype=np.int32)
        
        # Incremental mean: add valid pixels
        mask = ~np.isnan(arr)
        avg[mask] += arr[mask]
        count[mask] += 1
        
        elapsed = time.perf_counter() - t0
        print(f'File {i} done, time: {elapsed:.2f}s')
        
        # Explicitly delete arr and ds_crop to free memory
        del arr, ds_crop, ds
    
    # Compute final mean
    avg[count > 0] /= count[count > 0]
    avg[count == 0] = np.nan
    
    total_time = time.perf_counter() - start_time
    print(f'TOTAL TIME = {total_time:.2f}s')
    
    return avg
```

```python
print('XARRAY')
start_time = time.perf_counter()
avg_xr = XR_composite(arrs_ds)
print(f'time: {time.perf_counter() - start_time}')

print('\n----------------\nINC 1')
avg_inc1 = INC_composite(arrs)

print('\n----------------\nINC 2')
avg_inc2 = composite_rasters(fns, max_x, max_y, min_x, min_y, dtype=np.float16)
```
