## 1. Retrieve the shapefiles

```python
# Find the filename
import os
import geopandas as gpd
dir = "/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CSMUB/DATA/DEM/STATIONS/"
files = os.listdir(dir)
i = 11
fn = f'{dir}{files[i]}/{files[i]}.shp'
print(fn)
```

```python
# Open up the file
shp = gpd.read_file(fn)
shp["geometry"]
shp["geometry"].iloc[0]
```

## 2. Convert coords to EASE grid

```python
shp_ease = shp["geometry"].to_crs("epsg:6933")
shp_ease.iloc[0]
```
