```python
import geopandas as gpd
import pandas as pd
import os
```

```python
gdf = gpd.read_file('/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CSMUB/DATA/SHAPEFILES/stationbasins.geojson')
print(gdf.head())
```

```python
# Define output directory
out_dir = '/global/scratch/users/cgerlein/fc_ecohydrology_scratch/CSMUB/DATA/SHAPEFILES'
os.makedirs(out_dir, exist_ok=True)

station_nums_err = [1258202, 1258400, 1258501, 1258503, 1291200, 2854500, 2964122, 2964130, 2966400, 2969151, 3102010, 3103010, 3103015, 3103150, 3103300, 3107010, 3107015, 
                    3458015, 3458030, 3458040, 3512010, 3512020, 3512025, 3512100, 3513010, 3513015, 3514010, 3514015, 3514800, 3617811, 4126350, 4149301, 5202070]

# Filter gdf for only those stations
gdf_filtered = gdf[gdf['grdc_no'].isin(station_nums_err)]

# Loop through each unique grdc_no
for grdc in gdf_filtered['grdc_no'].unique():
    # Filter rows for this grdc_no
    gdf_subset = gdf[gdf['grdc_no'] == grdc]
    
    # Create a clean filename (ensure it's an integer without decimals)
    grdc_str = str(int(grdc)) if not pd.isna(grdc) else 'unknown'
    
    # Make a subfolder named by station number
    subfolder = os.path.join(out_dir, grdc_str)
    os.makedirs(subfolder, exist_ok=True)
    
    # Define the shapefile path inside that folder
    shapefile_path = os.path.join(subfolder, f"{grdc_str}.shp")
    
    # Save shapefile (this writes .shp, .dbf, .shx, .prj, .cpg, etc.)
    gdf_subset.to_file(shapefile_path)

    print(f"Saved {shapefile_path}")
```
